```python
import numpy as np
```

```python
A = np.array([[3,2],[4,1]])
print(A)
```

```python
autovalores, autovectores = np.linalg.eig(A)
```

```python
print(autovalores)
print(autovectores)
```

```python
A_calc = autovectores.dot(np.diag(autovalores)).dot(np.linalg.inv(autovectores))
```

```python
print(A_calc)
```

```python
B = np.array([[3,2],[2,3]])
print(B)
```

```python
print(B == B.T)
```

```python
autovalores, autovectores = np.linalg.eig(B)
print(autovalores)
print(autovectores)
```

```python
B_calc = autovectores.dot(np.diag(autovalores)).dot(autovectores.T)
print(B_calc)
```
