```python
A = np.array([[3,7],[5,2]])
print(A)
```

```python
print("circulo unitario:")
graficar_matriz(np.array([[1, 0], [0, 1]]), vectorCols=["red","blue"],limites={'x':(-1.5, 1.5), 'y': (-1.5,1.5)})
# plt.show()
```

```python
print("cirvulo unitario transformado:")
graficar_matriz(A, vectorCols=["red","blue"],limites={'x':(-8, 8), 'y': (-8,8)})
# plt.show()
```

```python
U, D, V = np.linalg.svd(A)
print(U)
```

```python
print("Primer rotacion V:")
graficar_matriz(V, vectorCols=["red","blue"],limites={'x':(-1.5, 1.5), 'y': (-1.5,1.5)})
```

```python
print("Escala (D):")
graficar_matriz(np.diag(D).dot(V), vectorCols=["red","blue"],limites={'x':(-9, 9), 'y': (-9,9)})
```

```python
print("Segunda rotacion U:")
graficar_matriz(U.dot(np.diag(D).dot(V)), vectorCols=["red","blue"],limites={'x':(-8, 8), 'y': (-8,8)})
```
